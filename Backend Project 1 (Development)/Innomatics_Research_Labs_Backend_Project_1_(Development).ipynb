```python
import re

def regex_matcher():
    test_string = input("Enter the test string: ")
    pattern = input("Enter the regex pattern: ")

    try:
        matches = re.findall(pattern, test_string)

        if matches:
            print("\nMatched strings:")
            for match in matches:
                print(match)
        else:
            print("\nNo matches found.")

    except re.error as e:
        print(f"\nInvalid regex pattern: {e}")

if __name__ == "__main__":
    regex_matcher()
```
