## ***Univariate Analysis***

```python
import plotly.express as px

score_counts = df["Score"].value_counts().sort_index().reset_index()
score_counts.columns = ["Score", "Student_Count"]

fig = px.bar(
    score_counts,
    x="Score",
    y="Student_Count",
    title="Score Frequency Distribution",
    labels={"Score": "Score (0–7)", "Student_Count": "Number of Students"},
    width=600,
    height=500,
)

fig.show()
```

**Observation form above distribution plot:**

The overall score distribution is **centered around the mid-range (3–5)**, indicating that most students achieved **moderate scores**. Very **few students scored at the extremes** (0–1 or 6–7), suggesting limited underperformance and fewer top scorers. Overall, the distribution appears **roughly balanced with a slight concentration toward average performance**, highlighting scope for targeted improvement to move more students into higher score bands


```python
import plotly.express as px

color_discrete_map = {
    "AI_ELITE_7": "teal",
    "AI_ELITE_6": "steelblue",
    "AI_ELITE_4": "royalblue"
}

batch_counts_df = df["Batch"].value_counts().reset_index()
batch_counts_df.columns = ["Batch_Name", "Student_Count"]

fig = px.bar(
    batch_counts_df,
    x="Batch_Name",
    y="Student_Count",
    color="Batch_Name",                     # 👈 IMPORTANT
    color_discrete_map=color_discrete_map,  # 👈 APPLY COLORS
    title="Number of Students per Batch",
    labels={"Batch_Name": "Batch", "Student_Count": "Student Count"},
    width=600,
    height=500,
)

fig.show()
```

**Observation from above Bar plots:**

The number of students across the three batches is **fairly balanced**, with **AI_ELITE_7** having a **slightly higher student count** compared to **AI_ELITE_6** and **AI_ELITE_4**. Overall, there is **no major imbalance in batch sizes**, indicating that comparisons of performance across batches are **fair and not biased by batch strength**.


## ***Bivariate Analysis***

```python
import plotly.express as px

color_discrete_map = {
    "AI_ELITE_7": "tomato",
    "AI_ELITE_6": "lightgreen",
    "AI_ELITE_4": "royalblue"
}

fig = px.box(
    df,
    x="Batch",
    y="Score",
    color="Batch",
    color_discrete_map=color_discrete_map,  # 👈 ADD THIS
    title="Score Distribution by Batch (Batch vs Score)",
    width=700,
    height=400,
)

fig.show()
```

**Observation from above boxplots:**

The batch **AI_ELITE_7** shows the **highest median score**, indicating stronger overall performance compared to the other batches. **AI_ELITE_6** and **AI_ELITE_4** have slightly lower median scores with a **wider spread**, suggesting greater variability in student performance. Overall, AI_ELITE_7 appears to be the **most consistent and better-performing batch**, while the other two batches show more mixed outcomes.


```python

import plotly.express as px

batch_avg = df.groupby("Batch")["Score"].mean().reset_index()

color_discrete_map = {
    "AI_ELITE_7": "lightcyan",
    "AI_ELITE_6": "lightgreen",
    "AI_ELITE_4": "lightblue"
}


fig = px.bar(
    batch_avg,
    x="Batch",
    y="Score",
    title="Average Score per Batch (Batch vs Average Score)",
    text_auto=".2f",
    color="Batch",
    color_discrete_map=color_discrete_map,
    width=600,
    height=500,
)

fig.show()
```

**Observation from above plots:**

The average score increases progressively across batches, with **AI_ELITE_7** achieving the **highest average score (~5.06)**, followed by **AI_ELITE_6 (~4.23)** and **AI_ELITE_4 (~3.79)**. This indicates a **clear performance gradient**, suggesting that students in higher-numbered batches are performing better on average.


```python
import plotly.express as px

color_discrete_map = {
    "AI_ELITE_7": "rosybrown",
    "AI_ELITE_6": "lightgreen",
    "AI_ELITE_4": "lightblue"
}


fig = px.histogram(
    df,
    x="Score",
    color="Batch",
    color_discrete_map=color_discrete_map,
    barmode="group",
    title="Score Distribution per Batch",
    width=700,
    height=500,
)

fig.show()
```

**Observation from above Stacked barplots:**

The score distribution varies noticeably across batches. **AI_ELITE_7** shows a higher concentration of students scoring in the **upper score range (5–7)**, indicating stronger overall performance. **AI_ELITE_6** is more evenly spread across mid-range scores, while **AI_ELITE_4** has a higher concentration around **lower to mid scores (2–4)**. This suggests a **performance progression across batches**, with AI_ELITE_7 consistently outperforming the others.


## ***Multivariate Analysis***

```python
import plotly.express as px

color_discrete_map = {
    "AI_ELITE_7": "cadetblue",
    "AI_ELITE_6": "lightcoral",
    "AI_ELITE_4": "lightsteelblue"
}


fig = px.violin(
    df,
    x="Batch",
    y="Score",
    color="Batch",
    color_discrete_map=color_discrete_map,
    box=True,
    points="all",
    title="Score Density & Spread Across Batches",
    width=700,
    height=500,
)

fig.show()
```

**Observation from above Violin plots:**

The violin plot shows clear differences in score distribution across batches. **AI_ELITE_7** has a higher concentration of scores in the **upper range (around 4–6)** with relatively less spread, indicating **strong and consistent performance**. **AI_ELITE_6** displays a wider spread of scores, suggesting **greater variability** among students. **AI_ELITE_4** is centered around **lower to mid-range scores (3–4)**, indicating comparatively weaker overall performance.


```python
import plotly.express as px

score_counts = (
    df.groupby(["Batch", "Score"])
    .size()
    .reset_index(name="Student_Count")
)


color_discrete_map = {
    "AI_ELITE_7": "darkgreen",
    "AI_ELITE_6": "lightcoral",
    "AI_ELITE_4": "darkblue"
}


fig = px.scatter(
    score_counts,
    x="Score",
    y="Batch",
    size="Student_Count",
    color="Batch",
    color_discrete_map=color_discrete_map,
    title="Score Concentration by Batch",
    size_max=40,
    width=700,
    height=500,
)

fig.show()
```

**Observation from above bubble chart:**

The bubble chart shows a clear shift in score concentration across batches. **AI_ELITE_7** has larger bubbles at **higher score values (5–7)**, indicating a higher concentration of top-performing students. **AI_ELITE_6** is mainly concentrated around **mid-range scores (4–5)**, while **AI_ELITE_4** shows larger concentrations at **lower to mid scores (3–4)**. Overall, the pattern highlights a **progressive improvement in performance from AI_ELITE_4 to AI_ELITE_7**.


```python
heatmap_data = pd.crosstab(df["Batch"], df["Score"])

fig = px.imshow(
    heatmap_data,
    text_auto=True,
    title="Heatmap: Batch vs Score Frequency",
    labels=dict(x="Score", y="Batch", color="Student Count"),
    color_continuous_scale="Teal",
    width=700,
    height=500,
)

fig.show()
```

**Observation from above Heatmap:**

* **AI_ELITE_4** has the highest concentration of students at **score 4**, indicating a strong mid-level performance but fewer high scorers.
* **AI_ELITE_6** shows a balanced distribution, with higher frequencies around **scores 4–5**, suggesting steady average performance.
* **AI_ELITE_7** is clearly skewed toward **higher scores (5–7)**, reflecting stronger overall outcomes and more top-performing students.
* Very **few low scores (0–1)** are observed across all batches, indicating minimal underperformance.
* Overall, the heatmap highlights a **progressive improvement in score distribution from AI_ELITE_4 to AI_ELITE_7**.
